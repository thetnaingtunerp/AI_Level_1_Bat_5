```python
import tensorflow as tf 

# Create a tensor
a = [[1, 2, 3], [4, 5, 6]]
tensor = tf.constant(a)

# Get the shape of the tensor
shape = tensor.shape

print('Tensor: ', tensor)
print('Tensor shape: ', shape)

```

```python
size = tf.size(tensor)
print("Tensor size: ", size)
```

```python
import tensorflow as tf
t1 = tf.constant([1, 2, 3, 4, 5, 6, 7])

#a slice from index 1 to 3
s1 = tf.slice(t1, begin=[1], size=[3])
print(s1)
s2 = t1[1:4]
```
