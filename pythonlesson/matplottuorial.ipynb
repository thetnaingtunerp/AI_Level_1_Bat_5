```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60]
y = [2,4,5,2,6,8]
z= [ 1,3,5,6,7,9]

plt.plot(x,y, marker='o' , linestyle='dotted', color='red')  #line 
plt.plot(x, z)

plt.title('this is Title')
plt.xlabel("this is X axis")
plt.ylabel('this is Y')

plt.show()

# 1,2,3,4,8
# <>, v,o, |, d, D
# Xx, pP, hH, *, s, 

```

```python
import matplotlib.pyplot as plt
name = ['mg mg', 'aung aung', 'ko ko', 'ma ma']
total = [20,30,25,45]

plt.bar(name, total ,color='yellow', width=0.2)

plt.show()
```

```python
import matplotlib.pyplot as plt
# scatter 
num1 = [2,4,6,8,10,12,14]
num2 = [1,3,15,7,9,17,13]

num3 = [2,3,4,12,5,4,7]
num4 = [6,3,2,5,8,12,3]
plt.scatter(num1,num2, color='red', s=[300,900,100,100,100,100,100])
plt.scatter(num1,num3, color='blue')
plt.scatter(num2,num4, color='#88c999')
plt.show()
```

```python
# pie chart
import matplotlib.pyplot as plt
num2 = [1,3,15,7,9]
lab = ['apple', 'orange', 'mango', 'pie-apple', 'banana']

plt.pie(num2, labels=lab)
plt.legend(title='fruit legend')
plt.title('this is pie chart')
plt.show()
```
