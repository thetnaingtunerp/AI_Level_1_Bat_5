```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
data = {
    'hour': [5,10,15,20,25,30],
    'score': [50,20,33,45,66,20]
}
df = pd.DataFrame(data)
X = df[['hour']]
y = df['score']

X_train, X_test, y_train, y_test=train_test_split(X, y, test_size = 0.2)


# model 
model = LinearRegression()
model.fit(X_train, y_train)

result =model.predict([[25]])
print(result)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression


shop= {
    'age': [12, 23,45,34,20,30,32],
    'min': [5, 10, 23,20, 4, 22, 7],
    'buy': [0,1,1,0,1,0,0]
}
df1 = pd.DataFrame(shop)

X = df1[['age', 'min']]
y = df1['buy']

X_train, X_test, y_train, y_test=train_test_split(X, y, test_size = 0.2)

model = LogisticRegression()
model.fit(X_train, y_train)

result = model.predict([[20, 5]])
print(result)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

data = {
    'name': ['mg mg', 'ma ma', 'aye aye', 'kyaw kyaw', 'su su'],
    'age': [20, 23, 20, np.nan, 25],
    'salary': [2000, 3200, np.nan, 3500, 3000]
}
df2 = pd.DataFrame(data)


model =SimpleImputer(strategy='mean')
model.fit(df2[['age', 'salary']])

df2[['age']] = model.transform(df2[['age']])
df2
```
