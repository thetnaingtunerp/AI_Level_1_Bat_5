```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.model_selection import train_test_split

df = pd.read_csv('dataset/wine.csv')
df.head()
df.shape
df.isnull().sum()
df = df.dropna()
df.isnull().sum()

df['type'].value_counts()
```

```python
df['is_white_wine'] = [1 if t == 'white' else 0 for t in df['type']]
df.drop('type', axis=1, inplace=True)
df.head()
```

```python
df['is_good_quality'] = [1 if q >= 6 else 0 for q in df['quality']]
df.drop('quality', axis=1, inplace=True)
df.head()
```
